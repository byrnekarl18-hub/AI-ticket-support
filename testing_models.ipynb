```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("data/clean/ticket_dataset.csv")
```

```python
df.head()
```

```python
X = df["ticket"]
y = df["priority"]
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2, random_state=11)
```

```python
# fist, I will test to simply tweak the (tf - idf) parameters i.e ngram range, max features
```

```python
v = TfidfVectorizer(
    ngram_range=(1,2),
    stop_words = "english",
    max_features = 5000
)
Xtfidf = v.fit_transform(X_train)
f = v.get_feature_names_out()
# print(f)
```

```python
from sklearn.linear_model import LogisticRegression

X_test_tfidf = v.transform(X_test)
model = LogisticRegression(max_iter=1000)
model.fit(Xtfidf, y_train)
y_pred = model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
# This result shows that a different ngram range of two connected words in addition of a smaller feature actually makes the model worse
```
